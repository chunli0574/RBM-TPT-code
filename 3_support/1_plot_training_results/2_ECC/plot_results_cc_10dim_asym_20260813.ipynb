```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
from tensorflow import keras
from scipy.stats import multivariate_normal as normal
import munch
import pylab
import sys


plot_gap = 0.5

for pace in pace_list:
    for a_str in a_str_list:
        for rho in rho_list:
            for theta in theta_list:
                b_h = 6.0*rho
                b_v = 1.5*rho
        
                file_name = f"RBMSteadyState_extendedCrissCross_none_{num_NN}_{num_NN}_{num_NN}_D4_{theta}_0.0_0.01_{a_str}_0.1_64_9000_dim=10_TD1_elu_100_addcost={addcost}_0.2_0.3_lr270000_eta={rho}_pricing=Y_delay=6.0_1.5"
                model_path = f"../../../1_NN_training/4_ECC_saved_NNs/{file_name}_grad"
                model = tf.saved_model.load(model_path)
        
                plt.rcParams.update({'font.size': 28,
                                   'text.usetex': True})
        
                font1 = {'family': 'Arial',
                        'weight': 'normal',
                        'size': 28}
        
        
                # plot single boundary control
        
        
                np.set_printoptions(precision=6)
                np.set_printoptions(suppress=True)
                np.set_printoptions(threshold=sys.maxsize)
        
                suffix = ""
                char2num = {str(i): i for i in range(10)}.__getitem__
        
                activation = "elu"
        
                cur_model_grad = keras.models.load_model("../../../1_NN_training/4_ECC_saved_NNs/" + file_name+"_grad")
        
                for bb in [b_h + b_v/2]: #[b_h + 0.5, b_h + 1.0]:
                    S_shrink = (np.minimum(bb,3)/3+1)/2
                    S_shrink2 = np.minimum(bb,3)/3
                    steps = 200
                    # upper_bound = 8
                    plt.rcParams["figure.figsize"] = (7.5,7.5)
                    plt.rcParams.update({'font.size': 28})
                    plt.rc('legend',fontsize=28)
                    x1list = np.linspace(-plot_gap, b_v + b_h + plot_gap, steps)
                    x2list = np.linspace(-plot_gap, b_v + b_h + plot_gap, steps)
                    x3list = np.array([bb])
                    x4list = np.array([bb])
                    x5list = np.array([bb])
                    x6list = np.array([bb])
                    x7list = np.array([bb])
                    x8list = np.array([bb])
                    x9list = np.array([bb])
                    x10list = np.array([bb])
        
                    X1, X2, X3, X4, X5, X6, X7, X8, X9, X10 = np.meshgrid(x1list, x2list, x3list, x4list, x5list, x6list, x7list, x8list, x9list, x10list)
        
                    X1_f = X1.flatten()
                    X2_f = X2.flatten()
                    X3_f = X3.flatten()
                    X4_f = X4.flatten()
                    X5_f = X5.flatten()
                    X6_f = X6.flatten()
                    X7_f = X7.flatten()
                    X8_f = X8.flatten()
                    X9_f = X9.flatten()
                    X10_f = X10.flatten()
        
                    points = np.concatenate((X1_f.reshape(1,-1), X2_f.reshape(1,-1), X3_f.reshape(1,-1), X4_f.reshape(1,-1), X5_f.reshape(1,-1), X6_f.reshape(1,-1), X7_f.reshape(1,-1), X8_f.reshape(1,-1), X9_f.reshape(1,-1), X10_f.reshape(1,-1)), axis=0).transpose()
        
                    cur_grad = cur_model_grad(points)
        
        
                    Z0 = cur_grad[:,0].numpy()
                    Z1 = cur_grad[:,1].numpy()
                    Z2 = cur_grad[:,2].numpy()
                    Z3 = cur_grad[:,3].numpy()
                    Z4 = cur_grad[:,4].numpy()
                    Z5 = cur_grad[:,5].numpy()
                    Z6 = cur_grad[:,6].numpy()
                    Z7 = cur_grad[:,7].numpy()
                    Z8 = cur_grad[:,8].numpy()
                    Z9 = cur_grad[:,9].numpy()
        
                    fig = plt.figure(figsize = [14, 7])
        
                    X1_mat = X1_f.reshape([len(x2list), len(x1list)])
                    X2_mat = X2_f.reshape([len(x2list), len(x1list)])
                    Z0_mat = Z0.reshape([len(x2list), len(x1list)])
                    Z1_mat = Z1.reshape([len(x2list), len(x1list)])
        
                    plt.subplot(1,2,1)
        
                    CS = plt.contourf(X1_mat, X2_mat, Z0_mat, np.arange(-0.1, 1.1, 0.1), cmap='jet')
                    plt.colorbar()
                    
                    x = np.linspace(0, (b_v+b_h)*S_shrink, steps)
                    # mask outside the state space (white); overlap the strips to avoid seams
                    x_mask = np.linspace(-plot_gap, b_v+b_h+plot_gap, steps)
                    plt.fill_between(x_mask, -plot_gap, np.maximum(0, x_mask - b_v), facecolor='white', zorder=1.5)
                    plt.fill_between(x_mask, (b_v+b_h)*S_shrink, b_v+b_h+plot_gap, facecolor='white', zorder=1.5)
                    plt.fill_betweenx([-plot_gap, b_v+b_h+plot_gap], -plot_gap, 0, facecolor='white', zorder=1.5)
                    plt.fill_betweenx([-plot_gap, b_v+b_h+plot_gap], (b_v+b_h)*S_shrink, b_v+b_h+plot_gap, facecolor='white', zorder=1.5)
                    # plt.plot(x, np.maximum(0,x-b_v),'--w', linewidth=2.5)
                    # plt.plot(x, np.ones(steps)*(b_v+b_h)*S_shrink, '--w', linewidth=2.5)
                    # plt.plot(np.ones(50)*0*S_shrink, np.linspace(0, b_h + b_v, 50)*S_shrink, '--w', linewidth=2.5)
                    # plt.plot(np.ones(50)*(b_v+b_h)*S_shrink, b_h*S_shrink2 + np.linspace(0, b_v, 50), '--w', linewidth=2.5)
                    plt.xlabel(r'$w_1$')
                    plt.ylabel(r'$w_2$')
                    plt.xlim([0, b_v + b_h])
                    plt.ylim([0, b_v + b_h])
                    plt.xticks([0, 2, 4, 6])
                    plt.yticks([0, 2, 4, 6])
                    plt.title(r'Gradient along $w_1$')
                    plt.tight_layout()
        
                    plt.subplot(1,2,2)
        
                    CS = plt.contourf(X1_mat, X2_mat, Z1_mat, np.arange(-0.1, 1.1, 0.1), cmap='jet')
                    plt.colorbar()
                    
                    x = np.linspace(0, (b_v+b_h)*S_shrink, steps)
                    # mask outside the state space (white); overlap the strips to avoid seams
                    x_mask = np.linspace(-plot_gap, b_v+b_h+plot_gap, steps)
                    plt.fill_between(x_mask, -plot_gap, np.maximum(0, x_mask - b_v), facecolor='white', zorder=1.5)
                    plt.fill_between(x_mask, (b_v+b_h)*S_shrink, b_v+b_h+plot_gap, facecolor='white', zorder=1.5)
                    plt.fill_betweenx([-plot_gap, b_v+b_h+plot_gap], -plot_gap, 0, facecolor='white', zorder=1.5)
                    plt.fill_betweenx([-plot_gap, b_v+b_h+plot_gap], (b_v+b_h)*S_shrink, b_v+b_h+plot_gap, facecolor='white', zorder=1.5)
                    # plt.plot(x, np.maximum(0,x-b_v),'--w', linewidth=2.5)
                    # plt.plot(x, np.ones(steps)*(b_v+b_h)*S_shrink, '--w', linewidth=2.5)
                    # plt.plot(np.ones(50)*0*S_shrink, np.linspace(0, b_h + b_v, 50)*S_shrink, '--w', linewidth=2.5)
                    # plt.plot(np.ones(50)*(b_v+b_h)*S_shrink, b_h*S_shrink2 + np.linspace(0, b_v, 50), '--w', linewidth=2.5)
                    plt.xlabel(r'$w_1$')
                    plt.ylabel(r'$w_2$')
                    plt.xlim([0, b_v + b_h])
                    plt.ylim([0, b_v + b_h])
                    plt.xticks([0, 2, 4, 6])
                    plt.yticks([0, 2, 4, 6])
                    plt.title(r'Gradient along $w_2$')
                    plt.tight_layout()

                    fig.suptitle(r'Slice: $w_3 = \cdots = w_{10} = b_h + b_v/2$', fontsize=36)
                    fig.tight_layout(rect=[0, 0, 1, 1.05])
        
        
                    plt.savefig(f"{figure_name}/ECC10_NN_gradient_upper.png")
                    plt.close()
```
