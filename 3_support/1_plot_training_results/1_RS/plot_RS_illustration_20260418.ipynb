```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
from tensorflow import keras
from scipy.stats import multivariate_normal as normal
import munch
import pylab
import sys
```

```python
from matplotlib.patches import Patch, FancyArrow
from matplotlib.legend_handler import HandlerPatch
from matplotlib.lines import Line2D
from matplotlib.ticker import ScalarFormatter

date_str = "20260418"
output_folder_name = f'png_elu_{date_str}'

plt.rcParams.update({'font.size': 25,
                   'text.usetex': True})

font1 = {'family': 'Arial',
        'weight': 'normal',
        'size': 25}

np.set_printoptions(precision=6)
np.set_printoptions(suppress=True)
np.set_printoptions(threshold=sys.maxsize)


# plot boundary control 

plt.figure(figsize = [7.5,7])

# used to draw shading

n = 1

steps = 200

eta = 8.0/9


# mask out the part of the contour outside the state space
t = np.arange(0,7.5*eta/n,1/100/n)
yt0 = np.ones(t.shape)*(-1)
yt1 = np.maximum(0, 2.0*t-10*eta/n)
yt2 = np.minimum(0.5*t+5*eta/n, 7.5*eta/n)
yt3 = np.ones(t.shape)*8.5/n

plt.fill_between(t,yt1,yt2, facecolor=[0.9,0.9,0.9], zorder = 2)
# plt.fill_between(t,yt2,yt3, facecolor='grey', zorder = 2)

# draw state space
x = np.linspace(0, 7.5*eta/n, steps, endpoint = True)
# x = np.arange(7.5*eta, step = 1 / steps, endpoint = True)
plt.plot(x, np.maximum(0,2*x-10*eta/n),'-k', linewidth=3)
plt.plot(x, np.minimum(7.5*eta/n,0.5*x+5*eta/n), '-k', linewidth=3)
plt.plot(np.ones(steps)*0, np.linspace(0, 5*eta/n, steps, endpoint = True), '-k', linewidth=3)
plt.plot(np.ones(steps)*7.5*eta/n, np.linspace(0, 2.5*eta/n, steps, endpoint = True) + 5*eta/n, '-k', linewidth=3)


plt.xlabel(r'$w_1$',font1)
plt.ylabel(r'$w_2$',font1)
# CS1.collections[0].set_label(r'$I_1$')
# CS2.collections[0].set_label(r'$I_2$')
# CS3.collections[0].set_label(r'$U_1$')
# #     CS4.collections[0].set_label(r'$U_3$')

move = 0

arrow = [[(-1.8 * eta /n, 3.5 * eta /n), (-0.2 * eta /n, 3.5 * eta /n)],
        [(3.5 * eta /n, -1.8 * eta /n), (3.5 * eta /n, -0.2 * eta /n)],
        [((4.215-move) * eta /n, (8.43-move) * eta /n), ((3.5-move) * eta /n, (7-move) * eta /n)],
        [((8.43-move) * eta /n, (4.215-move) * eta /n), ((7-move) * eta /n, (3.5-move) * eta /n)]]
        # [(0.4, 6.2), (2, 6.2)]
color_set = ['red','blue','green','darkorange','red']
legend_set = [r'$L_1$', r'$L_2$', r'$U_1$', r'$U_2$']

# Add arrows
for i in range(4):
    start = arrow[i][0]
    end = arrow[i][1]
    plt.annotate(
        '', xy=end, xytext=start,
        arrowprops=dict(arrowstyle='-|>', color=color_set[i], linewidth = 3)
    )

labels = [r'$F_1$', r'$F_2$', r'$F_3$', r'$F_4$', r'$F_5$', r'$F_6$']
positions = [
    (0.5 * eta / n, 2.5 * eta / n),  # edge 1 midpoint
    ((2.5+0.4) * eta / n, (6.25-0.4) * eta / n),  # edge 2 midpoint
    (6 * eta / n, (7.5-0.5) * eta / n),  # edge 3 midpoint
    ((7.5-0.5) * eta / n, 6 * eta / n),  # edge 4 midpoint
    ((6.25-0.4) * eta / n, (2.5+0.4) * eta / n),  # edge 5 midpoint
    (2.5 * eta / n, 0.5 * eta / n)  # point 6 (adjusted)
]
for label, (x, y) in zip(labels, positions):
    plt.text(x, y, label, fontsize = 24, ha='center', va='center')

        
# Custom legend

# Custom legend
custom_legend = [
    Patch(facecolor=[0.9,0.9,0.9], edgecolor='black', label=r'$\mathcal{W}$'),
    Line2D([0], [0], color=color_set[0], marker='None', markersize=10, linestyle='-', linewidth = 3, label=legend_set[0]),
    Line2D([0], [0], color=color_set[1], marker='None', markersize=10, linestyle='-', linewidth = 3, label=legend_set[1]),
    Line2D([0], [0], color=color_set[2], marker='None', markersize=10, linestyle='-', linewidth = 3, label=legend_set[2]),
    Line2D([0], [0], color=color_set[3], marker='None', markersize=10, linestyle='-', linewidth = 3, label=legend_set[3]),
]


# custom_legend = [
#     Patch(facecolor=[0.7,0.7,0.7], edgecolor='black', label=r'State space'),
#     # Line2D([0], [0], color=color_set[0], marker='>', markersize=10, linestyle='-', label=legend_set[0]),
#     # Line2D([0], [0], color=color_set[1], marker='>', markersize=10, linestyle='-', label=legend_set[1]),
#     # Line2D([0], [0], color=color_set[2], marker='>', markersize=10, linestyle='-', label=legend_set[2]),
#     # Line2D([0], [0], color=color_set[3], marker='>', markersize=10, linestyle='-', label=legend_set[3]),
# ]

plt.legend(handles=custom_legend, loc = 'lower right', fontsize = 22)

# # Custom legend
# custom_legend = [
#     Patch(facecolor='grey', edgecolor='black', label=r'State space')
# ]

# plt.legend(handles=custom_legend, loc = 'lower right', fontsize = 22)


# plt.title(r'Boundary control' + ', pricing = ' + pp + ', r = ' + discount_rate, font1)
# plt.legend(loc = 'lower right', fontsize = 22)
plt.xlim([-1.9*eta/n,10.5*eta/n])
plt.ylim([-1.9*eta/n,9*eta/n])
plt.xticks([0/n,2/n,4/n,6/n,8/n])
plt.yticks([0/n,2/n,4/n,6/n,8/n])

# formatter = ScalarFormatter(useMathText=True)
# formatter.set_powerlimits((-2, 2))  # specify when to switch to scientific notation
# plt.gca().xaxis.set_major_formatter(formatter)
# plt.gca().yaxis.set_major_formatter(formatter)

# plt.ticklabel_format(style='scientific', axis='x', scilimits=(-2, 2))
# plt.ticklabel_format(style='scientific', axis='y', scilimits=(-2, 2))


plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().xaxis.set_ticks_position('bottom')
plt.gca().yaxis.set_ticks_position('left')
plt.tight_layout()
plt.grid(False)
# plt.axis('off')

# Save the figure as a PDF file
plt.savefig(f'{output_folder_name}/RS_state_space_{date_str}.pdf', format='pdf', bbox_inches='tight', pad_inches=0.05)
```
